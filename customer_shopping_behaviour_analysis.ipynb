```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
#since the value of the review rating as some null values taking median across the category
#and and assigning to the null values of the category respectively
```

```python
df.isnull().sum()
```

```python
#since the columns are having spaces and uppercase to make it confined by using snake casing (lowercase with underscore)
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
labels = ['young adult','adult','middle-aged','senior']
#trying to create new column called age_group to map accordingly to age
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
#NOTE : this something ranging as given it may change the namiing scnerio if we change the order of the labels
```

```python
df['frequency_of_purchases'].unique()#gets the unique or distinct values of the column
```

```python
#create new column Purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,   
    'Weekly' : 7,     
    'Annually' : 365,  
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
#step : 1 connect to postgresql 
from sqlalchemy import create_engine

username = "postgres" #default user
password = "sa123" #  the password setted during the installation process
host = "localhost" # if running loally
port = "5432" # default port for postgresql
database = "customer_behaviour" #the DB created in pgadmin 

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

#step : 2 load dataframe into postgresql
table_name = "customer"
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"data successfully loaded into table '{table_name}' in database {database} ")
```
